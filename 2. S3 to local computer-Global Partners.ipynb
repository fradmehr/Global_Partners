```python
import boto3
import os

aws_access_key_id = ""
aws_secret_access_key = ""
region_name = "us-east-1"
bucket_name = "rds-sqlserver-csv-export"


# Downloads directory
downloads_dir = os.path.join(os.path.expanduser("~"), "Downloads")
os.makedirs(downloads_dir, exist_ok=True)

# Initialize S3 client
s3 = boto3.client(
    "s3",
    aws_access_key_id=aws_access_key_id,
    aws_secret_access_key=aws_secret_access_key,
    region_name=region_name,
)

# List objects in bucket
response = s3.list_objects_v2(Bucket=bucket_name)

if "Contents" not in response:
    print("Bucket is empty or you have no read permissions.")
else:
    for obj in response["Contents"]:
        key = obj["Key"]  # example: "date_dim/date_dim_part-0000-abc.csv"

        # Skip "folders" (S3 prefixes) and non-csv files
        if key.endswith("/") or not key.lower().endswith(".csv"):
            continue

        # -----------------------------------------------------
        # Use folder name as the final CSV filename
        # -----------------------------------------------------
        folder_name = key.split("/")[0]         # "date_dim"
        clean_filename = folder_name + ".csv"   # "date_dim.csv"

        local_path = os.path.join(downloads_dir, clean_filename)

        print(f"Downloading {key} → {clean_filename}")
        s3.download_file(bucket_name, key, local_path)

    print("\nAll CSV files downloaded successfully!")

```
